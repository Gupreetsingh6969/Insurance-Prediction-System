```python
from scipy.stats import pearsonr

selected_features=[
     'age', 'is_female', 'bmi', 'children', 'is_smoker', 'charges',
       'region_northwest', 'region_southeast', 'region_southwest'
       ]

correlation={
     feature:pearsonr(df_cleaned[feature],df_cleaned['charges'])[0]
      for feature in selected_features

 }

correlation_df=pd.DataFrame(list(correlation.items()),columns=['Feature','Pearson Correlation'])
correlation_df.sort_values(by='Pearson Correlation', ascending=False)
```

## Saving the Trained Model and Scaler

It's a good practice to save trained machine learning models and preprocessing objects (like our `StandardScaler`) so that they can be reused later without the need to retrain them. We'll use the `pickle` library for this, which serializes Python objects into a byte stream that can be stored in a file.

```python
import pickle

# Save the trained Linear Regression model
with open('linear_regression_model.pkl', 'wb') as file:
    pickle.dump(model, file)
print("Linear Regression model saved as 'linear_regression_model.pkl'")

# Save the fitted StandardScaler
with open('scaler.pkl', 'wb') as file:
    pickle.dump(scaler, file)
print("Scaler saved as 'scaler.pkl'")
```
